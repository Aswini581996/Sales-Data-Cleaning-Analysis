 ## Sales Data Cleaning

### Business Problem
The retail company is preparing its monthly sales performance report, but the raw sales data contains missing values, duplicate records, inconsistent formats, incorrect data types, invalid numerical values, and potentially incorrect Sales and Profit values. If these issues are not resolved before analysis, management may make decisions using inaccurate revenue, profitability, product, regional, and salesperson information. Your task is to act as a Data Analyst and convert the raw sales file into a reliable, consistent, and analysis-ready dataset that can be safely used for reporting and future EDA

## Objective of the data
1. Data Loading & Initial Inspection — Load the raw CSV into Pandas and inspect its shape, columns, sample records, data types, and overall structure.
2. Missing Values & Blank Data — Identify missing values, empty strings, and whitespace-only values, then determine an appropriate treatment for each affected field.
3. Duplicate Records & Order IDs — Detect exact duplicate rows and duplicate Order_ID values, investigate the affected records, and decide which records should remain.
4. Data Type & Format Correction — Correct incorrect or mixed data types and standardize Order_Date, Quantity, Unit_Price, Discount, Sales, and Profit into appropriate formats.
5. Text Standardization — Clean and standardize Region, Product, Category, and Salesperson values by handling spaces, capitalization differences, and inconsistent labels.
6. Invalid Value Detection — Identify negative, impossible, or out-of-range values in Quantity, Unit_Price, Discount, Sales, and Profit and decide how each issue should be resolved.
7. Business Rule Validation — Check whether Sales is logically consistent with Quantity, Unit_Price, and Discount, and investigate suspicious Profit values using an appropriate business rule.

## Import modules

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1.Data Loading and Initial Inspection

```python
data = pd.read_csv("Sales Data Cleaning Task.csv")
print("Dataset Shape:", data.shape)
print("\nColumn Names:",data.columns)
print("\nSample Records:",data.head())
print("\nData Types:",data.dtypes)
print("\nOverall Structure:",data.info())


```

## 2.Missing Values & Blank Data

```python
print("Missing Values:")
print(data.isnull().sum())
```

```python
print(data.apply(lambda col: col.astype("string").str.strip().eq("").sum()))
```

```python
data["Order_Date"] = pd.to_datetime(data["Order_Date"], errors="coerce")
data["Quantity"] = pd.to_numeric(data["Quantity"], errors="coerce")
data["Unit_Price"] = pd.to_numeric(data["Unit_Price"], errors="coerce")
data["Sales"] = pd.to_numeric(data["Sales"], errors="coerce")
data["Profit"] = pd.to_numeric(data["Profit"], errors="coerce")
```

```python
data["Quantity"] = data["Quantity"].fillna(data["Quantity"].median())
data["Unit_Price"] = data["Unit_Price"].fillna(data["Unit_Price"].median())
data["Sales"] = data["Sales"].fillna(data["Sales"].median())
data["Profit"] = data["Profit"].fillna(data["Profit"].median())
data["Region"] = data["Region"].fillna("Unknown")
data["Product"] = data["Product"].fillna("Unknown")
data["Salesperson"] = data["Salesperson"].fillna("Unknown")
data["Order_Date"] = data["Order_Date"].fillna("Unknown")
print("Missing Values:")
print(data.isnull().sum())
data = data.apply(lambda col: col.astype(str).str.strip().replace("", "Unknown"))
print(data.apply(lambda col: col.astype("string").str.strip().eq("").sum()))
```

## 3.Duplicate Records & Order IDs

```python
print("Exact Duplicate Records:")
print(data[data.duplicated(keep=False)])
print("\nRecords with Duplicate Order_ID:")
print(data[data.duplicated(subset=["Order_ID"], keep=False)])
```

```python
cleaned_data = data.drop_duplicates()
cleaned_data.to_csv("Cleaned_Sales_Data.csv", index=False)
print("\nExact duplicates removed.")
print("Remaining rows:", len(data))
```

## 4.Data Type & Format Correction

```python
data["Order_Date"] = pd.to_datetime(data["Order_Date"], errors="coerce")
data["Quantity"] = pd.to_numeric(data["Quantity"], errors="coerce")
data["Unit_Price"] = pd.to_numeric(data["Unit_Price"], errors="coerce")
print(data[["Order_Date", "Quantity", "Unit_Price","Discount", "Sales", "Profit"]].dtypes)
data.to_csv("Cleaned_Sales_Data.csv", index=False)
```

## 5.Text Standardization 

```python
columns = ["Region", "Product", "Category", "Salesperson"]
for col in columns:
    data[col] = data[col].astype("string").str.strip().str.title()
data["Region"] = data["Region"].replace({
    "North Region": "North",
    "South Region": "South",
    "West Region": "West",
    "East Region": "East"
})
data["Category"] = data["Category"].replace({
    "Electronics ": "Electronics",
    "Accessories": "Accessories",
    "Wearables": "Wearables",
})
data["Product"] = data["Product"].replace({
    "Headphones ": "Headphones",
    "Keyboard": "Keyboard",
    "Laptop": "Laptop",
    "Monitor ": "Monitor",
    "Mouse ": "Mouse",
    "Smartphone ": "Smartphone",
    "Smartwatch ": "Smartwatch",
    "Tablet ": "Tablet",
})
data["Salesperson"] = data["Salesperson"].replace({
    "Arun": "Arun",
    "Divya": "Divya",
    "Karthick": "Karthick",
    "Meena ": "Meena",
    "Priya": "Priya",
    "Rahul": "Rahul",
    "Sneha": "Sneha",
    "Vijay": "Vijay",
})
for col in columns:
    print(f"\n{col} values:")
    print(data[col].unique())
cleaned_data.to_csv("Cleaned_Sales_Data.csv", index=False)
    
```

## 6. Invalid Value Detection

```python
columns = ["Quantity", "Unit_Price", "Discount", "Sales", "Profit"]

for col in columns:
    data[col] = pd.to_numeric(
        data[col].astype(str).str.replace(r"[^\d.-]", "", regex=True),
        errors="coerce"
    )
print("\nQuantity less than or equal to zero:")
print(data[data["Quantity"] <= 0][["Order_ID", "Quantity"]])

print("\nUnit Price less than or equal to zero:")
print(data[data["Unit_Price"] <= 0][["Order_ID", "Unit_Price"]])

print("\nInvalid Discount values:")
print(data[(data["Discount"] < 0) | (data["Discount"] > 100)]
      [["Order_ID", "Discount"]])

print("\nNegative Sales:")
print(data[data["Sales"] < 0][["Order_ID", "Sales"]])

print("\nNegative Profit:")
print(data[data["Profit"] < 0][["Order_ID", "Profit"]])
```

## 7.Business Rule Validation

```python
data["Expected_Sales"] = (
    data["Quantity"] * data["Unit_Price"] *
    (1 - data["Discount"] / 100)
)
data["Sales_Difference"] = (
    data["Sales"] - data["Expected_Sales"]
).abs()
print("Suspicious Sales Records:")
print(data[data["Sales_Difference"] > 1]
      [["Order_ID", "Quantity", "Unit_Price",
        "Discount", "Sales", "Expected_Sales"]])
print("\nSuspicious Profit Records:")
print(data[data["Profit"] > data["Sales"]]
      [["Order_ID", "Sales", "Profit"]])
print("\nNegative Profit Records:")
print(data[data["Profit"] < 0]
      [["Order_ID", "Sales", "Profit"]])
```

## 8.Create the Cleaned Dataset

```python
raw_data = pd.read_csv("Sales Data Cleaning Task.csv")
cleaned_data = raw_data.copy()
cleaned_data = data.replace(r"^\s*$", pd.NA, regex=True)
cleaned_data = data.drop_duplicates()
text_columns = ["Region", "Product", "Category", "Salesperson"]
for col in text_columns:
    data[col] = data[col].astype("string").str.strip().str.title()
data["Order_Date"] = pd.to_datetime(data["Order_Date"], errors="coerce", dayfirst=True)
numeric_columns = ["Quantity", "Unit_Price", "Discount", "Sales", "Profit"]
for col in numeric_columns:
        data[col] = pd.to_numeric(
        data[col].astype(str).str.replace(r"[^\d.-]", "", regex=True),
        errors="coerce"
    )
data["Quantity"] = data["Quantity"].fillna(data["Quantity"].median())
data["Unit_Price"] = data["Unit_Price"].fillna(data["Unit_Price"].median())
print("Original dataset shape:", raw_data.shape)
print("Cleaned dataset shape:", data.shape)
print("\nMissing values after cleaning:")
print(data.isnull().sum())
cleaned_data.to_csv("Cleaned_Sales_Data.csv", index=False)
print("\nCleaned dataset saved successfully.")
```
